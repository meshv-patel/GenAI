```python
# ReLU

def relu(x):
    return np.maximum(0, x)

x = np.array([-3, -1, 0, 2, 5])

y = relu(x)

print("Input: ", x)
print("After ReLU: ", y)
```

```python
# Sigmoid

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.array([-3, -1, 0, 2, 5])

y = sigmoid(x)

print("Input: ", x)
print("After Sigmoid: ", y)
```

```python
# Softmax
# Converts score into probabilities, majorly used in multi-class classification

def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

scores = np. array([2.0, 1.0, 0.1])

probabilities = softmax (scores)
print("Scores: ", scores)
print("Probabilities: ", probabilities)
print("Sum of probabilities: ", np.sum(probabilities))
```

```python
# Manual MLP
# Architecture: Input -> Linear -> ReLU -> Linear -> Softmax

np.random.seed(42)

x = np.array([1.0, 2.0, 3.0, 4.0])

# First layer : 4 inputs features -> 5 hidden neurons 
W1 = np.random.randn(4, 5)
b1 = np.random.randn(5)

# Second layer : 5 hidden neurons -> 3 classes
W2 = np.random.randn(5, 3)
b2 = np.random.randn(3)

# Forward Pass
z1 = x @ W1 + b1
a1 = relu(z1)

z2 = a1 @ W2 + b2
output = softmax(z2.flatten())

print("Input shape:", x.shape)
print("Hidden output shape:", a1.shape)
print("Final output probabilities:", output)
print("Predicted class:", np.argmax(output))
```

```python
# Simple 2-D convolution
# No padding, stride = 1

image = np.array([
    [1, 2, 3, 0],
    [4, 5, 6, 1],
    [7, 8, 9, 2],
    [1, 0, 1, 3]
])
kernel = np.array([
    [1, 0], 
    [0, -1]
])

image_height, image_width = image. shape
kernel_height, kernel_width = kernel. shape

output_height = image_height - kernel_height + 1
output_width = image_width - kernel_width + 1

output = np.zeros((output_height, output_width))

for i in range (output_height): 
    for j in range (output_width):
        region = image[i:i+kernel_height, j:j+kernel_width]
        output[i, j] = np.sum(region * kernel)

print("Input image: ", image)

print("Kernel: ", kernel)

print("Convolution output: ", output)
```

```python
# Max Pooling
# Pool size = 2x2
# Stride = 2

feature_map = np.array([
    [1, 3, 2, 4],
    [5, 6, 1, 2],
    [7, 8, 9, 0],
    [1, 2, 3, 4],
])

pool_size = 2
stride = 2

output_height = feature_map.shape[0] // 2
output_width = feature_map.shape[1] // 2

pooled = np.zeros( (output_height, output_width))
for i in range(output_height): 
    for j in range (output_width):
        row_start = i * stride
        col_start = j * stride

        region = feature_map[
            row_start: row_start+pool_size, 
            col_start: col_start+pool_size,
        ]

        pooled[i, j] = np.max(region)

print("Feature map: ", feature_map)

print("After max pooling: ", pooled)
```

```python
# Manual RNN Cell
#h_t = tanh (x_t W_xh + h_previous W_hh + b)

np.random.seed(42)

sequence_length = 5
input_dim = 3
hidden_dim = 4

# Sequence input
X = np. random.randn(sequence_length, input_dim)

# Weights
W_xh = np.random.randn(input_dim, hidden_dim)
W_hh = np.random.randn(hidden_dim, hidden_dim)
b = np.random.randn(hidden_dim)

# Initial hidden state
h = np.zeros(hidden_dim)

print("Initial hidden state:", h)

for t in range(sequence_length):
    x_t = X[t]

    h = np.tanh(x_t @W_xh + h @W_hh + b)
    
    print ("Time step:", t) 
    print ("Hidden state:" , h)
```

```python
# Creating a Small Dataset
# Binary classification

np.random.seed(42)

# 100 samples, each with 2 features
X = np.random.rand(100, 2)

# Rule:
# If x1 + x2 › 0, class 1
# Else class 0
y = (X[:, 0] + X[:, 1] > 0). astype(int)

print("Input shape:", X.shape)
print("Target shape:", y. shape)

print("First 5 inputs:")
print(X[:5])

print("First 5 labels:")
print(y[:5])
```

```python
# Train-test split

num_samples = X.shape[0]

indices = np.arange(num_samples)

np.random.shuffle(indices)

train_size = int(0.8 * num_samples)

train_indices = indices[:train_size]
test_indices = indices[train_size:]

X_train = X[train_indices]
y_train = y[train_indices]

X_test = X[test_indices]
y_test = y[test_indices]

print("Train input shape:", X_train.shape)
print("Train target shape:", y_train.shape)
print("Test input shape:", X_test.shape)
print("Test target shape:", y_test. shape)
```

```python
# Accuracy Function

def accuracy(y_true, y_pred):
    correct = np.sum(y_true == y_pred)
    total = len(y_true)
    return correct / total

y_true = np.array([0, 1, 1, 0, 1])
y_pred = np.array([0, 1, 0, 0, 1])

acc = accuracy(y_true, y_pred)

print ("Accuracy: ", acc)
```

```python
# Logistic Regression from Scratch Using NumPy
# This is a single neuron binary classifier

np.random.seed(42)

# Dataset
X = np.random.randn(200, 2)
y = (X[:, 0] + X[:, 1] > 0). astype (int)
# Convert y shape from (200,) to (200, 1)
y = y. reshape (-1, 1)

# Initialize parameters
input_dim = 2

W = np.random.randn(input_dim, 1) * 0.01
b = 0.0

learning_rate = 0.1
epochs = 1000

def sigmoid (x):
    return 1 / (1 + np.exp(-x))

for epoch in range (epochs):
    
    # Forward pass
    logits = X @W + b
    y_pred = sigmoid (logits)

    # Binary cross entropy loss
    loss = -np. mean( 
        y * np.log(y_pred + 1e-8) + 
        (1 - y) * np.log(1 - y_pred + 1e-8)
    )
# Gradients
error = y_pred - y

dW = X.T @ error / X.shape[0]
db = np.mean(error)

# Parameter update
W = W - learning_rate * dW
b = b - learning_rate * db
if epoch% 100== 0:
    predictions = (y_pred >= 0.5).astype(int)
    acc = np.mean(predictions == y)
    print (f"Epoch {epoch}, Loss: {loss:.4f}, Accuracy: {acc: .4f}")

print("Final weights:")
print(W)

print("Final bias:")
print(b)
```

```python
# Prediction Using Trained Model

new_samples = np.array([
        [2.0, 1.0],
        [-1.0, -2.0],
        [1.5, -2.0]
    ])

logits = new_samples @W + b

probabilities = sigmoid(logits)
predictions = (probabilities >= 0.5).astype(int)

print("New samples: ")
print(new_samples)

print("Probabilities:")
print (probabilities)

print("Predicted classes:")
print (predictions)
```
